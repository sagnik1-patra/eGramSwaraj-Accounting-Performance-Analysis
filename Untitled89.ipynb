```python
import os
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import yaml
import h5py

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")


# =============================================================================
# 2. FILE PATHS
# =============================================================================

BASE_DIR = r"C:\Users\sagni\Downloads\eGramSwaraj Accounting Performance Analysis"

CSV_PATH = os.path.join(
    BASE_DIR,
    "egramswaraj_accounting_20_21.csv"
)

PKL_PATH = os.path.join(
    BASE_DIR,
    "egramswaraj_model.pkl"
)

H5_PATH = os.path.join(
    BASE_DIR,
    "egramswaraj_model.h5"
)

YAML_PATH = os.path.join(
    BASE_DIR,
    "egramswaraj_config.yaml"
)

JSON_PATH = os.path.join(
    BASE_DIR,
    "egramswaraj_metadata.json"
)


# Create output directory if it does not exist
os.makedirs(BASE_DIR, exist_ok=True)


# =============================================================================
# 3. LOAD DATASET
# =============================================================================

print("=" * 80)
print("LOADING DATASET")
print("=" * 80)

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}\n"
        "Please check the file path."
    )

df = pd.read_csv(CSV_PATH)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)

print("\nColumns:")
for column in df.columns:
    print("-", column)


# =============================================================================
# 4. DATASET PREVIEW
# =============================================================================

print("\n" + "=" * 80)
print("DATASET PREVIEW")
print("=" * 80)

print(df.head())

print("\nDataset information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())


# =============================================================================
# 5. CLEAN COLUMN NAMES
# =============================================================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)


print("\nCleaned columns:")

for column in df.columns:
    print("-", column)


# =============================================================================
# 6. REMOVE DUPLICATE RECORDS
# =============================================================================

initial_rows = len(df)

df = df.drop_duplicates().reset_index(drop=True)

removed_duplicates = initial_rows - len(df)

print("\nDuplicate rows removed:", removed_duplicates)
print("Remaining rows:", len(df))


# =============================================================================
# 7. HANDLE MISSING VALUES
# =============================================================================

numeric_columns = [
    "onboard_egspi",
    "pv_initiated",
    "pv_approved",
    "year_closed"
]


for column in numeric_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    df[column] = df[column].fillna(
        df[column].median()
    )


# Clean categorical columns

categorical_columns = [
    "state_name",
    "district_name"
]


for column in categorical_columns:

    df[column] = (
        df[column]
        .fillna("unknown")
        .astype(str)
        .str.strip()
        .str.lower()
    )


# =============================================================================
# 8. CREATE NORMALIZED PERFORMANCE INDICATORS
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PERFORMANCE INDICATORS")
print("=" * 80)


performance_features = [
    "onboard_egspi",
    "pv_initiated",
    "pv_approved",
    "year_closed"
]


normalized_features = []


for column in performance_features:

    new_column = column + "_normalized"

    min_value = df[column].min()
    max_value = df[column].max()

    if max_value != min_value:

        df[new_column] = (
            (df[column] - min_value)
            /
            (max_value - min_value)
        )

    else:

        df[new_column] = 0.0

    normalized_features.append(new_column)


print("\nNormalized features created:")

for column in normalized_features:
    print("-", column)


# =============================================================================
# 9. CREATE ACCOUNTING PERFORMANCE SCORE
# =============================================================================

# Weights:
#
# onboard_egspi = 20%
# pv_initiated  = 25%
# pv_approved   = 30%
# year_closed   = 25%
#
# PV approval receives slightly higher importance because it represents
# successful progression of the accounting/payment process.


df["performance_score"] = (

    df["onboard_egspi_normalized"] * 0.20

    +

    df["pv_initiated_normalized"] * 0.25

    +

    df["pv_approved_normalized"] * 0.30

    +

    df["year_closed_normalized"] * 0.25

)


# Convert score from 0-1 to 0-100

df["performance_score"] = (
    df["performance_score"] * 100
)


print("\nPerformance score created.")

print(
    df[
        [
            "state_name",
            "district_name",
            "performance_score"
        ]
    ].head()
)


# =============================================================================
# 10. CREATE PERFORMANCE CLASS
# =============================================================================

# Quantile-based classification is used because the raw indicator values
# may be highly skewed.
#
# Bottom 33%  -> Low
# Middle 33%  -> Medium
# Top 33%     -> High


q1 = df["performance_score"].quantile(0.33)
q2 = df["performance_score"].quantile(0.66)


print("\nPerformance thresholds:")
print("Low/Medium threshold :", q1)
print("Medium/High threshold:", q2)


def classify_performance(score):

    if score <= q1:
        return "Low"

    elif score <= q2:
        return "Medium"

    else:
        return "High"


df["performance_class"] = (
    df["performance_score"]
    .apply(classify_performance)
)


print("\nPerformance class distribution:")

print(
    df["performance_class"]
    .value_counts()
)


# =============================================================================
# 11. ENCODE TARGET
# =============================================================================

class_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}


df["performance_class_encoded"] = (
    df["performance_class"]
    .map(class_mapping)
)


# =============================================================================
# 12. SELECT MODEL FEATURES
# =============================================================================

# State and district codes are excluded because they are identifiers.
#
# Normalized indicators are used for classification.


feature_columns = [
    "onboard_egspi_normalized",
    "pv_initiated_normalized",
    "pv_approved_normalized",
    "year_closed_normalized"
]


X = df[feature_columns].copy()

y = df["performance_class_encoded"].copy()


print("\nFeature matrix shape:", X.shape)
print("Target shape:", y.shape)


# =============================================================================
# 13. TRAIN TEST SPLIT
# =============================================================================

print("\n" + "=" * 80)
print("TRAIN TEST SPLIT")
print("=" * 80)


# Check whether stratification can safely be used

class_counts = y.value_counts()

print("\nClass counts:")
print(class_counts)


if len(class_counts) > 1 and class_counts.min() >= 2:

    stratify_value = y

else:

    stratify_value = None


X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=stratify_value
)


print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# =============================================================================
# 14. STANDARDIZE FEATURES
# =============================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train
)


X_test_scaled = scaler.transform(
    X_test
)


print("\nFeature scaling completed.")


# =============================================================================
# 15. TRAIN RANDOM FOREST CLASSIFIER
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING RANDOM FOREST MODEL")
print("=" * 80)


model = RandomForestClassifier(

    n_estimators=300,

    max_depth=None,

    min_samples_split=2,

    min_samples_leaf=1,

    random_state=42,

    class_weight="balanced",

    n_jobs=-1
)


model.fit(
    X_train_scaled,
    y_train
)


print("\nRandom Forest training completed.")


# =============================================================================
# 16. MODEL PREDICTION
# =============================================================================

y_pred = model.predict(
    X_test_scaled
)


accuracy = accuracy_score(
    y_test,
    y_pred
)


print("\nModel Accuracy:")
print(f"{accuracy * 100:.2f}%")


# =============================================================================
# 17. CLASSIFICATION REPORT
# =============================================================================

labels_present = sorted(
    list(
        set(y_test.unique())
        |
        set(np.unique(y_pred))
    )
)


reverse_class_mapping = {
    0: "Low",
    1: "Medium",
    2: "High"
}


target_names = [
    reverse_class_mapping[label]
    for label in labels_present
]


report = classification_report(

    y_test,
    y_pred,

    labels=labels_present,

    target_names=target_names,

    zero_division=0,

    output_dict=True
)


print("\nClassification Report:")

print(
    classification_report(

        y_test,
        y_pred,

        labels=labels_present,

        target_names=target_names,

        zero_division=0
    )
)


# =============================================================================
# 18. CONFUSION MATRIX
# =============================================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=labels_present
)


print("\nConfusion Matrix:")
print(cm)


# =============================================================================
# 19. FEATURE IMPORTANCE
# =============================================================================

feature_importance = dict(

    zip(
        feature_columns,
        model.feature_importances_.tolist()
    )

)


print("\nFeature Importance:")

for feature, importance in sorted(
    feature_importance.items(),
    key=lambda x: x[1],
    reverse=True
):

    print(
        f"{feature}: {importance:.4f}"
    )


# =============================================================================
# 20. SAVE PKL FILE
# =============================================================================

print("\n" + "=" * 80)
print("SAVING PKL FILE")
print("=" * 80)


model_package = {

    "model": model,

    "scaler": scaler,

    "features": feature_columns,

    "class_mapping": class_mapping,

    "reverse_class_mapping": reverse_class_mapping,

    "performance_weights": {

        "onboard_egspi": 0.20,

        "pv_initiated": 0.25,

        "pv_approved": 0.30,

        "year_closed": 0.25
    },

    "performance_thresholds": {

        "low_medium": float(q1),

        "medium_high": float(q2)
    },

    "accuracy": float(accuracy)
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        model_package,
        file
    )


print("\nPKL saved successfully:")
print(PKL_PATH)


# =============================================================================
# 21. SAVE H5 FILE
# =============================================================================

print("\n" + "=" * 80)
print("SAVING H5 FILE")
print("=" * 80)


with h5py.File(
    H5_PATH,
    "w"
) as h5_file:

    # ---------------------------------------------------------
    # Model information
    # ---------------------------------------------------------

    h5_file.attrs["model_name"] = (
        "RandomForestClassifier"
    )

    h5_file.attrs["project_name"] = (
        "AI-Based eGramSwaraj Accounting Performance Analysis"
    )

    h5_file.attrs["accuracy"] = (
        float(accuracy)
    )


    # ---------------------------------------------------------
    # Feature names
    # ---------------------------------------------------------

    string_dtype = h5py.string_dtype(
        encoding="utf-8"
    )


    h5_file.create_dataset(

        "feature_names",

        data=np.array(
            feature_columns,
            dtype=object
        ),

        dtype=string_dtype
    )


    # ---------------------------------------------------------
    # Class labels
    # ---------------------------------------------------------

    class_names = [
        "Low",
        "Medium",
        "High"
    ]


    h5_file.create_dataset(

        "class_names",

        data=np.array(
            class_names,
            dtype=object
        ),

        dtype=string_dtype
    )


    # ---------------------------------------------------------
    # Scaler information
    # ---------------------------------------------------------

    scaler_group = h5_file.create_group(
        "scaler"
    )


    scaler_group.create_dataset(
        "mean",
        data=scaler.mean_
    )


    scaler_group.create_dataset(
        "scale",
        data=scaler.scale_
    )


    scaler_group.create_dataset(
        "var",
        data=scaler.var_
    )


    # ---------------------------------------------------------
    # Random Forest information
    # ---------------------------------------------------------

    model_group = h5_file.create_group(
        "random_forest"
    )


    model_group.attrs["n_estimators"] = (
        model.n_estimators
    )


    model_group.attrs["n_features"] = (
        model.n_features_in_
    )


    model_group.attrs["n_classes"] = (
        len(model.classes_)
    )


    model_group.create_dataset(
        "classes",
        data=model.classes_
    )


    model_group.create_dataset(
        "feature_importances",
        data=model.feature_importances_
    )


    # ---------------------------------------------------------
    # Store performance thresholds
    # ---------------------------------------------------------

    threshold_group = h5_file.create_group(
        "performance_thresholds"
    )


    threshold_group.attrs["low_medium"] = (
        float(q1)
    )


    threshold_group.attrs["medium_high"] = (
        float(q2)
    )


    # ---------------------------------------------------------
    # Store confusion matrix
    # ---------------------------------------------------------

    h5_file.create_dataset(
        "confusion_matrix",
        data=cm
    )


print("\nH5 saved successfully:")
print(H5_PATH)


# =============================================================================
# 22. SAVE YAML CONFIGURATION
# =============================================================================

print("\n" + "=" * 80)
print("SAVING YAML FILE")
print("=" * 80)


yaml_config = {

    "project": {

        "name":
            "AI-Based eGramSwaraj Accounting Performance Analysis",

        "description":
            "District-level accounting performance classification "
            "using eGramSwaraj 2020-21 accounting indicators.",

        "dataset":
            CSV_PATH
    },


    "dataset_information": {

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1]),

        "financial_year":
            "2020-2021"
    },


    "features": feature_columns,


    "original_performance_features": [

        "onboard_egspi",

        "pv_initiated",

        "pv_approved",

        "year_closed"
    ],


    "target": {

        "column":
            "performance_class",

        "classes": {

            "Low": 0,

            "Medium": 1,

            "High": 2
        }
    },


    "performance_score": {

        "weights": {

            "onboard_egspi": 0.20,

            "pv_initiated": 0.25,

            "pv_approved": 0.30,

            "year_closed": 0.25
        },

        "thresholds": {

            "low_medium":
                float(q1),

            "medium_high":
                float(q2)
        }
    },


    "model": {

        "algorithm":
            "RandomForestClassifier",

        "n_estimators":
            300,

        "random_state":
            42,

        "class_weight":
            "balanced"
    },


    "training": {

        "test_size":
            0.20,

        "scaling":
            "StandardScaler"
    },


    "performance": {

        "accuracy":
            float(accuracy)
    },


    "output_files": {

        "pkl":
            PKL_PATH,

        "h5":
            H5_PATH,

        "yaml":
            YAML_PATH,

        "json":
            JSON_PATH
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.dump(

        yaml_config,

        file,

        allow_unicode=True,

        sort_keys=False
    )


print("\nYAML saved successfully:")
print(YAML_PATH)


# =============================================================================
# 23. SAVE JSON METADATA
# =============================================================================

print("\n" + "=" * 80)
print("SAVING JSON FILE")
print("=" * 80)


json_metadata = {

    "project_name":
        "AI-Based eGramSwaraj Accounting Performance Analysis "
        "and District-Level Classification",


    "dataset": {

        "file_name":
            "egramswaraj_accounting_20_21.csv",

        "file_path":
            CSV_PATH,

        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1]),

        "financial_year":
            "2020-2021"
    },


    "model": {

        "name":
            "Random Forest Classifier",

        "accuracy":
            float(accuracy),

        "accuracy_percentage":
            round(float(accuracy * 100), 2),

        "number_of_trees":
            int(model.n_estimators)
    },


    "features":
        feature_columns,


    "target": {

        "name":
            "performance_class",

        "mapping":
            class_mapping
    },


    "performance_score_weights": {

        "onboard_egspi":
            0.20,

        "pv_initiated":
            0.25,

        "pv_approved":
            0.30,

        "year_closed":
            0.25
    },


    "performance_thresholds": {

        "low_medium":
            float(q1),

        "medium_high":
            float(q2)
    },


    "feature_importance":
        feature_importance,


    "confusion_matrix":
        cm.tolist(),


    "classification_report":
        report,


    "saved_files": {

        "pkl":
            os.path.basename(PKL_PATH),

        "h5":
            os.path.basename(H5_PATH),

        "yaml":
            os.path.basename(YAML_PATH),

        "json":
            os.path.basename(JSON_PATH)
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(

        json_metadata,

        file,

        indent=4,

        ensure_ascii=False
    )


print("\nJSON saved successfully:")
print(JSON_PATH)


# =============================================================================
# 24. VERIFY SAVED FILES
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING OUTPUT FILES")
print("=" * 80)


output_files = [

    PKL_PATH,

    H5_PATH,

    YAML_PATH,

    JSON_PATH
]


for file_path in output_files:

    if os.path.exists(file_path):

        file_size = (
            os.path.getsize(file_path)
            /
            1024
        )

        print(
            f"\n[OK] {os.path.basename(file_path)}"
        )

        print(
            f"     Size: {file_size:.2f} KB"
        )

    else:

        print(
            f"\n[ERROR] File not found: {file_path}"
        )


# =============================================================================
# 25. TEST LOADING PKL MODEL
# =============================================================================

print("\n" + "=" * 80)
print("TESTING SAVED PKL MODEL")
print("=" * 80)


with open(
    PKL_PATH,
    "rb"
) as file:

    loaded_package = pickle.load(
        file
    )


loaded_model = loaded_package["model"]

loaded_scaler = loaded_package["scaler"]


test_sample = X.iloc[[0]]


test_sample_scaled = (
    loaded_scaler.transform(
        test_sample
    )
)


test_prediction = (
    loaded_model.predict(
        test_sample_scaled
    )[0]
)


predicted_class = (
    reverse_class_mapping[
        int(test_prediction)
    ]
)


print("\nSample district:")
print(df.iloc[0]["district_name"])

print("State:")
print(df.iloc[0]["state_name"])

print("Predicted performance class:")
print(predicted_class)


# =============================================================================
# 26. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("PROJECT FILE GENERATION COMPLETED")
print("=" * 80)


print(
    f"""
Project:
AI-Based eGramSwaraj Accounting Performance Analysis

Dataset:
{CSV_PATH}

Dataset Records:
{len(df)}

Features Used:
{feature_columns}

Target:
performance_class

Classes:
0 = Low
1 = Medium
2 = High

Model:
Random Forest Classifier

Accuracy:
{accuracy * 100:.2f}%

Files Generated:

1. {PKL_PATH}

2. {H5_PATH}

3. {YAML_PATH}

4. {JSON_PATH}

All requested files were generated successfully.
"""
)
```
